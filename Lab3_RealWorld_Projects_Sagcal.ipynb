```python
# Equipment Diagnostic System

from util import LAST_NAME, SEED_NUM, FAVORITE_ARTIST
import random

print("=== Equipment Diagnostic System ===")
print("Student:", LAST_NAME, "\nSeed Number:", SEED_NUM, "\nFavorite Artist:", FAVORITE_ARTIST)

random.seed(len(LAST_NAME)+SEED_NUM+len(FAVORITE_ARTIST))
data=[random.randint(-5,15) for _ in range(5)]
print("Generated Equipment Data:",data)

valid=[(x,"VALID") for x in data]
print("Validation Results:",valid)

def classify(x):
    if x<0: return "FAULT"
    if x<=5: return "NORMAL"
    if x<=10: return "WARNING"
    return "CRITICAL"

diag=[(x,classify(x)) for x,_ in valid]
print("Diagnostic Results:",diag)

print("Execution Log: Process complete")
print("Final Output:",{c:sum(1 for _,y in diag if y==c) for c in ["FAULT","NORMAL","WARNING","CRITICAL"]})
```

```python
# Recursive Fault Trace

from util import LAST_NAME, SEED_NUM, FAVORITE_ARTIST

print("=== Fault Trace Program ===")
print("Student:", LAST_NAME, "\nSeed Number:", SEED_NUM, "\nFavorite Artist:", FAVORITE_ARTIST)

fault=len(LAST_NAME)+SEED_NUM+len(FAVORITE_ARTIST)
print("Generated Fault Data:",fault)

def trace(n,c=1):
    print("Recursive Trace",c,":",n)
    if n<=1: return "Resolved",c
    return trace(n-1,c+1)

result,calls=trace(fault)
print("Number of Recursive Calls:",calls)
print("Execution Log: Process complete")
print("Final Output:",result)
```

```python
# Intelligent Equipment Monitoring Pipeline

from util import LAST_NAME, SEED_NUM, FAVORITE_ARTIST
import random

print("=== Intelligent Equipment Monitoring Pipeline ===")
print("Student:", LAST_NAME, "\nSeed Number:", SEED_NUM, "\nFavorite Artist:", FAVORITE_ARTIST)

random.seed(len(LAST_NAME)+SEED_NUM+len(FAVORITE_ARTIST))
data=[random.randint(-10,20) for _ in range(8)]
print("Generated Telemetry Data:",data)

valid=[x for x in data if isinstance(x,int)]
invalid=[x for x in data if not isinstance(x,int)]
print("Valid/Invalid Results:",{"Valid":valid,"Invalid":invalid})

abnormal=[x for x in valid if x>10]
def recurse(v):
    print("Recursive check:",v)
    if v<=0: return "Resolved"
    return recurse(v-1)

report={"Processed":len(valid)+len(invalid),"Valid":len(valid),
        "Invalid":len(invalid),"Abnormal":len(abnormal),
        "Status":"CRITICAL" if abnormal else "NORMAL",
        "RecursiveAnalysis":recurse(abnormal[0]) if abnormal else "No abnormal detected"}

print("Processed Results:",report)
print("Execution Log: Process complete")
print("Final Output:",report)
```
